Rock Paper Scissor Game

```python
import random
```

```python
game = ['r', 'p', 's']

while True:
  computer = random.choice(game)
  choice = input("\nRock, Paper, or Scissors ? press(r/p/s): ").lower()
  emojis = {'r' : '✊', 'p': '🖐️', 's': '✌️'}

  if choice not in game:
    print("Invalid Choice!")
    continue

  print(f"You choose: {emojis[choice]}")
  print(f"Computer choose: {emojis[computer]}")

  if choice == computer:
    print("Game Tied!")
  elif(
      (choice == 's' and computer == 'p'),
      (choice == 'r' and computer == 's'),
      (choice == 'p' and computer == 'r')
  ):
    print("You Win!")
  else:
    print("You Lose!")

  retry = input('Continue? (y/n): ').lower()
  if retry == 'n':
    break
```
