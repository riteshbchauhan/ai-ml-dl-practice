Number Guessing Game

```python
import random
```

```python
while True:
  r = random.randint(1, 100)
  try:
    c = int(input("Guess the number between 1 to 100 [type: 0(zero) for exit]: "))

    if r == c:
      print("Great, you guess the right number")
    elif c == 0:
      print("Exit")
      break
    elif c < 0 or c > 100:
      print("Enter between 1 to 100")
    elif r != c:
      print(f'Wrong Choice!\nComputer: {r} and You: {c}\nBetter Luck Next Time!')
  except ValueError:
    print("Invalid Input! [Input contain only 1 to 100 number]")
```
