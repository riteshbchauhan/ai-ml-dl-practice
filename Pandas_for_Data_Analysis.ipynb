```python
df1 = pd.DataFrame({
    "name" : ["Harry", "Rohan", "Shivamn"],
    "marks": [90, 80, 85]
})

df2 = pd.DataFrame({
    "name" : ["Ram", "Mohamad", "Cebestian"],
    "marks": [99, 20, 85]
})

pd.concat([df1, df2])
```

```python
df3 = pd.DataFrame({
    "name" : ["Harry", "Rohan", "Shivamn"],
    "rno" : [1, 5,6]
})

pd.merge(df1, df3, on='name')
```
