##### Array Reshaping

```python
arr = np.arange(12)
print("Original Array: ", arr)

reshaped = arr.reshape((3,4))
print("Reshaped Array: \n", reshaped)

flatten = reshaped.flatten()
print("Flatten Array: ", flatten)

# ravel (returns view, instead of copy)
raveled = reshaped.ravel()
print("Raveled Array: ", raveled)

# Transpose
transpose = reshaped.T
print("Transposed Array: \n", transpose)
```

#### Fancy indexing VS np.where

```python
indices = [0,2,4]
print(numbers[indices])

where_result = np.where(numbers > 5)
print(where_result)
print("NP where: ", numbers[where_result])
```

```python
condition_array = np.where(numbers > 7 , numbers * 3, numbers)
print("Condition array: ", condition_array)
```

##### Array compatibility

```python
original = np.array([[1,2], [3,4]])
new_row = np.array([[5,6]])

with_new_row = np.vstack((original, new_row)) #vstack add new row
print("Original : \n", original)
print("Added new row: \n", with_new_row)

new_col = np.array([[7], [8]])
with_new_col = np.hstack((original, new_col))
print("Added new column: \n", with_new_col)
```

#### Advance operation with Business Examples

```python
# schema: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000], # Paradise Biryani
    [2, 120000, 140000, 160000, 190000], # Beijing Bites
    [3, 200000, 230000, 260000, 300000], # Pizaa Hub
    [4, 180000, 210000, 240000, 270000], # Burger Point
    [5, 160000, 185000, 285000, 230000]  # Chai Point
])
```

```python
import matplotlib.pyplot as plt # Visulization
```

```python
print("====== Zomoto Sales Analysis =====")
print("\n Sales data shape: ", sales_data.shape)
print("\n Sample data for 1st 3 restarunt: \n", sales_data[:3])
```

```python
# total sales per year

print(np.sum(sales_data, axis=0))
yearly_total = np.sum(sales_data[:, 1:], axis = 0)
print(yearly_total)
```

```python
# Minimum sales per rest.

min_sales = np.min(sales_data[:, 1:], axis=1)
print(min_sales)
```

```python
# Maximum sales per year

max_sales = np.max(sales_data[:, 1:], axis= 0)
print(max_sales)
```

```python
# average sales per restarant

avg_sales = np.average(sales_data[:, 1:], axis = 1)
print(avg_sales)
```

```python
# Cumilative sales

cumsum = np.cumsum(sales_data[:, 1:], axis = 1)
print(cumsum)
```

```python
plt.figure(figsize = (8,6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumlative sales across all rest.")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
