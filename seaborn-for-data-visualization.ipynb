```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
p = sns.load_dataset('penguins')
p.head()
```

```python
# scatter plot

sns.scatterplot(data = p, x = 'flipper_length_mm', y='body_mass_g', hue = 'island', palette = 'Dark2')
sns.set_context("notebook") # paper, notebook, talk, poster
```

```python
sns.scatterplot(data = p, x = "species", y = 'body_mass_g', hue ='island', style = 'sex')
```

```python
# stripplot

sns.stripplot(data = p, x = 'species', y = 'body_mass_g', hue = 'island', jitter = True)
```

```python
sns.stripplot(data = p, x = 'species', y = 'body_mass_g', hue = 'island', dodge = True, jitter = True)
```

```python
# swarmplot

sns.swarmplot(data = p, x = 'species', y = 'body_mass_g', hue = 'island')
```

```python
# histogram 

sns.histplot(data = p, x = 'body_mass_g', hue = 'sex')
```

```python
# regression plot

sns.regplot(data = p, x = 'body_mass_g', y = 'flipper_length_mm')
```

```python
sns.regplot(data = p, x = 'body_mass_g', y = 'flipper_length_mm', scatter = False)
```

```python
# line plot

sns.lineplot(data = p, x = 'body_mass_g', y = 'flipper_length_mm', hue = 'sex')
```

```python
# joint plot

sns.jointplot(data = p, x = 'body_mass_g', y ='flipper_length_mm')
```

```python
sns.jointplot(data = p, x = 'body_mass_g', y ='flipper_length_mm', hue = 'sex')
```

```python
# bar plot

sns.barplot(data = p, x = 'species', y = 'body_mass_g', hue = 'sex')
```

```python
# Count plot

sns.countplot(data = p, x = 'species', hue = 'island')
```

```python
# boxplot

sns.boxplot(data = p, x = 'species', y = 'body_mass_g', hue = 'sex')
```

```python
# violin plot

sns.violinplot(data = p, x = 'species', y = 'body_mass_g', hue  = 'sex')
```

```python
sns.violinplot(data = p, x = 'species', y = 'body_mass_g', hue  = 'sex', split = True)
```

```python
# KDE plot

sns.kdeplot(data = p, x ='body_mass_g', hue = 'species', palette = 'Dark2', fill = True)
```

```python
p.columns
```

```python
# heatmap

columns = ["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"]
sns.heatmap(data = p[columns].corr(), annot = True, cmap = 'coolwarm')
```

```python
# Rug plot

sns.rugplot(data = p, y = "body_mass_g", hue = "species", palette = "Set2", height = 0.2)
```

```python
# pair plot

sns.pairplot(data = p, hue = "sex", palette = 'Set1', diag_kind = "kde")
```

```python
# pair grid -> subplot grid for plotting pairwise relationships in a dataset

graph = sns.PairGrid(data = p, hue = 'sex', palette = "Set1")
graph.map_upper(sns.scatterplot) # upper tringle
graph.map_lower(sns.kdeplot) # lower triangle
graph.map_diag(sns.histplot) # diagnol
graph.add_legend()
```
