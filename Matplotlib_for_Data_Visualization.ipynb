### Anatomy of Figure (LOOK THE **OUTPUT**)

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.patches import Circle
from matplotlib.patheffects import withStroke
from matplotlib.ticker import AutoMinorLocator, MultipleLocator

royal_blue = [0, 20/256, 82/256]


# make the figure

np.random.seed(19680801)

X = np.linspace(0.5, 3.5, 100)
Y1 = 3+np.cos(X)
Y2 = 1+np.cos(1+X/0.75)/2
Y3 = np.random.uniform(Y1, Y2, len(X))

fig = plt.figure(figsize=(7.5, 7.5))
ax = fig.add_axes((0.2, 0.17, 0.68, 0.7), aspect=1)

ax.xaxis.set_major_locator(MultipleLocator(1.000))
ax.xaxis.set_minor_locator(AutoMinorLocator(4))
ax.yaxis.set_major_locator(MultipleLocator(1.000))
ax.yaxis.set_minor_locator(AutoMinorLocator(4))
ax.xaxis.set_minor_formatter("{x:.2f}")

ax.set_xlim(0, 4)
ax.set_ylim(0, 4)

ax.tick_params(which='major', width=1.0, length=10, labelsize=14)
ax.tick_params(which='minor', width=1.0, length=5, labelsize=10,
               labelcolor='0.25')

ax.grid(linestyle="--", linewidth=0.5, color='.25', zorder=-10)

ax.plot(X, Y1, c='C0', lw=2.5, label="Blue signal", zorder=10)
ax.plot(X, Y2, c='C1', lw=2.5, label="Orange signal")
ax.plot(X[::3], Y3[::3], linewidth=0, markersize=9,
        marker='s', markerfacecolor='none', markeredgecolor='C4',
        markeredgewidth=2.5)

ax.set_title("Anatomy of a figure", fontsize=20, verticalalignment='bottom')
ax.set_xlabel("x Axis label", fontsize=14)
ax.set_ylabel("y Axis label", fontsize=14)
ax.legend(loc="upper right", fontsize=14)


# Annotate the figure

def annotate(x, y, text, code):
    # Circle marker
    c = Circle((x, y), radius=0.15, clip_on=False, zorder=10, linewidth=2.5,
               edgecolor=royal_blue + [0.6], facecolor='none',
               path_effects=[withStroke(linewidth=7, foreground='white')])
    ax.add_artist(c)

    # use path_effects as a background for the texts
    # draw the path_effects and the colored text separately so that the
    # path_effects cannot clip other texts
    for path_effects in [[withStroke(linewidth=7, foreground='white')], []]:
        color = 'white' if path_effects else royal_blue
        ax.text(x, y-0.2, text, zorder=100,
                ha='center', va='top', weight='bold', color=color,
                style='italic', fontfamily='monospace',
                path_effects=path_effects)

        color = 'white' if path_effects else 'black'
        ax.text(x, y-0.33, code, zorder=100,
                ha='center', va='top', weight='normal', color=color,
                fontfamily='monospace', fontsize='medium',
                path_effects=path_effects)


annotate(3.5, -0.13, "Minor tick label", "ax.xaxis.set_minor_formatter")
annotate(-0.03, 1.0, "Major tick", "ax.yaxis.set_major_locator")
annotate(0.00, 3.75, "Minor tick", "ax.yaxis.set_minor_locator")
annotate(-0.15, 3.00, "Major tick label", "ax.yaxis.set_major_formatter")
annotate(1.68, -0.39, "xlabel", "ax.set_xlabel")
annotate(-0.38, 1.67, "ylabel", "ax.set_ylabel")
annotate(1.52, 4.15, "Title", "ax.set_title")
annotate(1.75, 2.80, "Line", "ax.plot")
annotate(2.25, 1.54, "Markers", "ax.scatter")
annotate(3.00, 3.00, "Grid", "ax.grid")
annotate(3.60, 3.58, "Legend", "ax.legend")
annotate(2.5, 0.55, "Axes", "fig.subplots")
annotate(4, 4.5, "Figure", "plt.figure")
annotate(0.65, 0.01, "x Axis", "ax.xaxis")
annotate(0, 0.36, "y Axis", "ax.yaxis")
annotate(4.0, 0.7, "Spine", "ax.spines")

# frame around figure
fig.patch.set(linewidth=4, edgecolor='0.5')
plt.show()
```

### PyPlot API

#### Univariate Analysis

```python
data = {
    "salary" : [20000, 25000, 20000, 10000, 15000, 12500, 52000, 33000, 44000, 35000]
}

df = pd.DataFrame(data)
```

##### Line chart

```python
plt.plot(df['salary'], color = 'red', marker = 'o', linestyle = '-.')
plt.grid()
plt.show()
```

##### Histogram

```python
plt.hist(df['salary'], bins = 5)
plt.grid()
plt.show()
```

##### Boxplot

```python
plt.boxplot(df['salary'])
plt.show()
```

##### Univariate: Categorical

```python
df['dept'] = ['HR', 'IT', 'Sales','Finance','Account'] * 2
```

```python
df
```

##### Pie chart

```python
count = df['dept'].value_counts()
count
```

```python
plt.pie(count, labels= count.index, autopct="%1.1f")
plt.show()
```

##### Count Plot

```python
plt.bar(count.index, count)
plt.show()
```

#### Bivariate : Numerical - Numerical

```python
df['age'] = [22, 25, 30, 25, 52, 20, 35, 63, 28, 19]
```

##### Scatter Plot

```python
plt.scatter(df['age'], df['salary'])
plt.show()
```

##### Line Chart

```python
sort_age = df.sort_values("age")
```

```python
plt.plot(sort_age['age'], df['salary'], color = 'red', marker = 'o', linestyle = "--")
plt.show()
```

##### Bar chart

```python
plt.bar(sort_age['age'], df['salary'])
plt.show()
```

#### Bivariate : Numerical - Categorical

```python
# ['HR', 'IT', 'Sales','Finance','Account']

hr_sal = df[df['dept'] == 'HR']['salary']
it_sal = df[df['dept'] == 'IT']['salary']
sales_sal = df[df['dept'] == 'Sales']['salary']
finance_sal = df[df['dept'] == 'Finance']['salary']
account_sal = df[df['dept'] == 'Account']['salary']
```

##### box plot

```python
plt.boxplot([hr_sal, it_sal, sales_sal, finance_sal, account_sal], labels=['HR', 'IT', 'Sales','Finance','Account'])
plt.show()
```

##### Pie Chart

```python
salary_by_dept = df.groupby('dept')['salary'].sum()

plt.pie(salary_by_dept, labels=salary_by_dept.index, autopct="%.2f")
plt.show()
```

#### Multivariate Analysis : 3 Numerical

```python
df['experience'] = [1, 1, 5, 3, 2.5, 1.8, 9, 2, 4, 10]
```

```python
df.head()
```

##### Bubble plot

```python
plt.scatter(df['age'], df['salary'], s = df['experience'] * 10, color = 'red')
```

#### Multivariate : 2 Numerical + 1 categorical

```python
plt.scatter(df['age'], df['salary'], c = df['dept'].map({'HR': 'yellow', 'IT' : 'red', 'Sales' : 'blue', 'Finance' : 'Orange', 'Account': 'green'}))
plt.show()
```

### Object Oriented API

```python
fig, axes = plt.subplots(1, 3, figsize= (15, 5))

# Line chart

axes[0].plot(sort_age['age'], df['salary'], color = 'red', marker = 'o')
axes[0].grid()
axes[0].set_title('Line Plot')
axes[0].set_xlabel('Age')
axes[0].set_ylabel('Salary')

# Histogram Chart

axes[1].hist(df['salary'], bins = 5, color = 'skyblue')
axes[1].set_title('Histogram Chart')
axes[1].set_xlabel('Salary')
axes[1].set_ylabel('Age')

# Box plot

axes[2].boxplot([hr_sal, it_sal, sales_sal, finance_sal, account_sal], labels=['HR', 'IT', 'Sales','Finance','Account'])
axes[2].set_title('Box Plot')
axes[2].set_xlabel('Salary')

# save the figure
plt.savefig('multipleplots.png')

plt.show
```

##### 3D Plot

```python
ax = plt.axes(projection = '3d')

ax.scatter(df['age'], df['salary'], df['experience'])
ax.set_xlabel('Age')
ax.set_ylabel('Salary')
ax.set_zlabel('Experience')
plt.show()
```
